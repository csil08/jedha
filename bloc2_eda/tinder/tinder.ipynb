# Tinder: speed dating data analysis

## Goal
This project aims at understanding **what drives mutual interest during speed dating** and which features influence a participant’s decision to agree to a second date.

## Dataset
Data was collected from participants in experimental speed dating events organized by Tinder and held between 2002 and 2004. Each participant had a four minute "first date" with every participant of the opposite sex. After each date, participants indicated whether they would like a second date and rated their date on six attributes: attractiveness, sincerity, intelligence, fun, ambition, and shared interests.

The dataset also includes questionnaire data collected at different points in the process, covering:
- Demographics
- Lifestyle information
- Dating habits
- Self-perception across key attributes
- Beliefs about what others find valuable in a partner

[Dataset](https://full-stack-assets.s3.eu-west-3.amazonaws.com/M03-EDA/Speed+Dating+Data.csv)

[Dataset Description](https://full-stack-assets.s3.eu-west-3.amazonaws.com/M03-EDA/Speed+Dating+Data+Key.doc)

## Methodology
This explanatory data analysis follows these steps:
- initial exploration and overview of the dataset;
- data cleaning; 
- general descriptive statistics;
- detailed descriptive analysis and visualisations about:
    - analysis of the valuable attributes in a partner, based on pre-experiment responses
    - comparison of actual behavior with stated preferences
    - assessment of participants' ability to gauge their own perceived value in the dating market
    - influence of the round order on the decision to request a second date

# 2. Data cleaning

This section is dedicated to:
- Dealing with missing values: checking and addressing missing values
- Performing data validation: updating data types, checking ranges for numerical data.

```python
# Display the number of missing values
df_count_na = df.isnull().sum().reset_index().rename(columns = {'index' : 'var_name', 0: "count_of_missing_values"})
df_count_na['proportion_of_missing_values'] = round(df_count_na['count_of_missing_values'] / len(df) * 100, 2)
df_count_na.sort_values(['proportion_of_missing_values'], 
                        ascending = False,
                        inplace = True)
df_count_na
```

```python
# Focus on variables which have a significant proportion of missing values (above 30%)

count_na_prct30 = len(df_count_na[df_count_na['proportion_of_missing_values'] >= 30])

print(f"{count_na_prct30} variables \
(accounting for {round(count_na_prct30/ len(df_count_na),2)*100} % of the total number of variables) \
have above 30% of missing observations")

# Visual representation 

fig = px.bar(df_count_na[df_count_na['proportion_of_missing_values'] >= 30], 
            x = 'var_name', 
            y = 'proportion_of_missing_values',
            title = "Percentage of missing values by variable (for variables with more than 30% of missing values)")
fig.update_layout(
    xaxis_title = 'Variable',
    yaxis_title = 'Percentage of missing values (%)',
    xaxis_tickangle = -45
)
fig.show()
```

Variables with a high percentage of missing values are often those related to the survey after the date experiment.

We decide to drop the variables with more than 70% of missing values: we are sure that we will not use them in the analysis.

```python
# Removing variables with more than 70% of missing values
cols_to_remove = df_count_na[df_count_na['proportion_of_missing_values'] >= 70]['var_name'].tolist()
print(f"{len(cols_to_remove)} columns with more than 70% missing values will be removed from the dataset.")
df = df.drop(columns = cols_to_remove)
print(df.shape)
```

```python
# Race
df['race'] = df['race'].map({1: 'Black/African American', 
                            2: 'European/Caucasian American',
                            3: 'Latino/Hispanic American',
                            4: 'Asian/Pacific Islander American',
                            5: 'Native American',
                            6: 'Other'})\
                        .astype('category')

df['race'].value_counts(dropna = False).reset_index()
```

### 4.1.1. Analysis for overall participants

Observations with missing values for all attributes' ratings are excluded.  
For the remaining missing values, we impute zero because the sum of the non missing attributes' ratings is theoretically equal to 100.

```python
# Deleting observations with missing values for all attributes' ratings
df_sp = df_sp.loc[~df_sp[attributes1.keys()].isna().all(axis=1)]
print(df_sp.shape)

# Imputation of the rest of the missing values: they are equal to 0
df_sp[list(attributes1.keys())] = df_sp[list(attributes1.keys())].fillna(0)
```

### Checking for the range of the attributes
The descriptive notice indicates that the rating scale differs differs between waves 6 to 9 and the rest of the waves:
- for waves 6 to 9 : the importance of each attribute in a potential date is rated on a scale of 1 to 10
- for the other waves : 100 points are distributed among the six attributes. Total points must equal 100.

We are going to check whether this is the case and deal with it.

```python
# Creating the score (sum of the attributes ratings)
df_sp['score1'] = df_sp['attr1_1'] + df_sp['sinc1_1'] + df_sp['intel1_1'] + df_sp['fun1_1'] + df_sp['amb1_1'] + df_sp['shar1_1']

# Dividing the dataset into two subsets
df_sp_wave_6_9 = df_sp.loc[(df['wave'] >= 6) & (df_sp['wave'] <= 9)]
df_sp_wave_rest = df_sp.loc[~((df['wave'] >= 6) & (df_sp['wave'] <= 9))]
```

```python
# Displaying the 5 first observations of waves 6-9
df_sp_wave_6_9[[*list(attributes1.keys()), "score1"]].head()
```

```python
# Descriptive statistics for wave 6 to 9
display(df_sp_wave_6_9[["score1", *list(attributes1.keys())]].describe().round(2))
```

```python
# Descriptive statistics for the rest of the waves
display(df_sp_wave_rest[["score1", *list(attributes1.keys())]].describe().round(2))
```

```python
display(df_sp[["score1", *list(attributes1.keys())]].describe().round(2))
max_threshold = 100
outlier_low = df_sp.loc[df_sp['score1'] < max_threshold]
outlier_high = df_sp.loc[df_sp['score1'] > max_threshold]
print(f"Observations out of the expected value:\n \
    {len(outlier_low)} observations display a total score below {max_threshold}.\n \
    {len(outlier_high)} observations display a total score above {max_threshold}.\n")
```

For the waves 6 to 9, contrary to what is stated in the dataset description, it seems that the attributes' scores have been rescaled since the sum of the ratings equals to 100.

For the rest of the waves (1-5 and 10-21), we detect observations out of the expected range for individual scores prior to the experiment, i.e. observations with a total score of attributes ratings higher than 100. 

We decide to **rescale the ratings for individual attributes** and **impute the percentage of each attribute in the total score**, so that the sum of the 6 ratings equals 100.


```python
# Rescaling the ratings attributes

# Determining the proportions of each attribute in each total score
df_sp.loc[:, 'attr1_1'] = df_sp['attr1_1'] / df_sp['score1'] * 100
df_sp.loc[:, 'sinc1_1'] = df_sp['sinc1_1'] / df_sp['score1'] * 100
df_sp.loc[:, 'intel1_1'] = df_sp['intel1_1'] / df_sp['score1'] * 100
df_sp.loc[:, 'fun1_1'] = df_sp['fun1_1'] / df_sp['score1'] * 100
df_sp.loc[:, 'amb1_1'] = df_sp['amb1_1'] / df_sp['score1'] * 100
df_sp.loc[:, 'shar1_1'] = df_sp['shar1_1'] / df_sp['score1'] * 100

df_sp.head()

# Removing the old score1 variable
df_sp = df_sp.drop(columns = ['score1'])

# Recreating the total score (sum of the 6 ratings)    
df_sp['score1'] = df_sp['attr1_1'] + df_sp['sinc1_1'] + \
    df_sp['intel1_1'] + df_sp['fun1_1'] + \
    df_sp['amb1_1'] + df_sp['shar1_1']

# Check statistics for the score variable
display(df_sp['score1'].describe().round(2))
```

```python
df_sp_describe = df_sp.describe().round(2).T
df_sp_describe['coeff_of_variation'] = round(df_sp_describe['std']/df_sp_describe['mean'], 2)
df_sp_describe['interquartile_range'] = round(df_sp_describe['75%']-df_sp_describe['25%'], 2)
df_sp_describe = df_sp_describe.T
df_sp_describe
```

Based on the descriptive statistics, we note that :
- attractiveness has the widest range of ratings (from 0 to 100), with a higher mean score and dispersion compared to the other attributes
- attractiveness, ambitiousness and shared interests scores have more variability than intelligence, fun and sincerity.

### 4.1.2. Analysis by gender

Let's see if the ranking of the attributes differ from the gender.

```python
# Descriptive statistics per gender
def describe_attrib_gender(input_df, listvar, gender_value):
    df_temp = input_df.loc[input_df['gender'] == gender_value][[*listvar, "gender"]].drop_duplicates().describe().round(2)
    df_temp_T = df_temp.T
    df_temp_T['coeff_of_variation'] = round(df_temp_T['std']/df_temp_T['mean'], 2)
    df_temp_T['interquartile_range'] = round(df_temp_T['75%']-df_temp_T['25%'], 2)
    return(df_temp_T.T)

print("Descriptive statistics for male")
display(describe_attrib_gender(df_sp, list(attributes1.keys()), "Male"))

print("Descriptive statistics for female")
display(describe_attrib_gender(df_sp, list(attributes1.keys()), "Female"))
```

## 4.4. Impact of the round order on the decision to ask for a second date

We would like to know if the round order of the date influences the decision to respond positively: is it better to be someone's first speed date of the night or their last?

### 4.4.1. Is it better of be someone's first speed date of the night or their last?

```python
cols_to_keep = ['order', 'round', 'dec']
df_first_last = df[cols_to_keep]
```

```python
df_first_last = df_first_last.loc[(df_rp['order'] == 1) | (df_first_last['order'] == df_first_last['round'])]
df_first_last.loc[df_first_last['order'] == 1, 'order_lib'] = "First"
df_first_last.loc[df_first_last['order'] == df_first_last['round'], 'order_lib'] = "Last"
display(df_first_last.head())
print("Shape of the subset:", df_first_last.shape)
```

```python
for lib in ['First', 'Last']:
    print("Descriptive statistics of decision for", lib.lower(), "order")
    display(df_first_last.loc[df_first_last['order_lib'] == lib]['dec'].describe().round(3))
```

```python
# Proportion of positive answer per order
df_first_last_answer = df_first_last.groupby(['order_lib'], observed = True, dropna = False).agg(
    nb_positive_answer = ("dec", "sum"),
    nb_answer = ("dec", "count")
).reset_index()
df_first_last_answer['positive_answer_rate'] = round(df_first_last_answer['nb_positive_answer']/df_first_last_answer['nb_answer'] * 100, 1)
display(df_first_last_answer)


fig = px.bar(
    df_first_last_answer,
    x = "order_lib",
    y = "positive_answer_rate",
    text_auto = '.1f',
    title = "Proportion of positive answer depending on the order (first or last)",
    color_discrete_sequence = ["lightseagreen"]
)

fig.update_layout(
    title_x=0.5,
    xaxis_title = "Order",
    yaxis_title = "Positive answer rate (%)",
)
fig.show()
```

Being on first position seems to be more favourable to get a second date: on average, the participant is likely to request a second date half of the time.
Being on last position is associated with a lower proportion of positive answer. One explanation could be social tiredness.

### 4.4.2. Does the order in the round influence the decision to respond positively?

```python
# Number of people met during each wave
df_round_wave = df_rp[['wave', 'round']].groupby('wave', dropna = False).agg(
    min_number_people_met = ("round", "min"),
    max_number_people_met = ("round", "max"),
).reset_index()
display(df_round_wave)

# Barplot of the minimum number of people met, per wave
fig = px.bar(df_round_wave, 
            x = 'wave', 
            y = 'min_number_people_met',
            title = "Minimum number of people met per wave")
fig.update_layout(title_x=0.5)
fig.show()
```

In each wave, the number of people met per participant is variable, so we cannot directly work with the `order` variable. 

Thus we compute for each participant the **progression in the round**, by dividing the order by the number of people met during the round. This variable varies from 0 to 1: a value near to 0 means the participant is at the beginning of the round whereas a value near to 1 means that he/she is towards the end.

```python
# Create a variable of progression during the experiment
df_rp.loc[:, 'round_progression'] = df_rp['order'] / df_rp['round']
```

```python
# Defining class of progression order

labels = ["[0 - 0.1[",
        "[0.1 - 0.2[",
        "[0.2 - 0.3[",
        "[0.3 - 0.4[",
        "[0.4 - 0.5[",
        "[0.5 - 0.6[",
        "[0.6 - 0.7[",
        "[0.7 - 0.8[",
        "[0.8 - 0.9[",
        ">= 0.9"
        ]

df_rp['progression_round_class'] = pd.cut(
    x = df_rp['round_progression'],
    bins = np.arange(0, 1.1, 0.1),
    labels = labels
)
```

```python
df_rp.head()
```

```python
# Displaying the number of observations per class
df_rp['progression_round_class'].value_counts(dropna = False)
```

```python
# Proportion of positive answer per class of progression
df_order_answer = df_rp.groupby(['progression_round_class'], observed = True, dropna = False).agg(
    nb_positive_answer = ("dec", "sum"),
    nb_answer = ("dec", "count")
).reset_index()
df_order_answer['positive_answer_rate'] = round(df_order_answer['nb_positive_answer']/df_order_answer['nb_answer'] * 100, 1)
df_order_answer
```

```python
fig = px.bar(
    df_order_answer,
    x = "progression_round_class",
    y = "positive_answer_rate",
    title = "Proportion of positive answer depending on the progression round",
    text_auto = '.1f',
    color_discrete_sequence = ["lightseagreen"]
)

fig.update_layout(
    title_x=0.5,
    xaxis_title = 'Round progression',
    yaxis_title = 'Percentage of positive answer (%)'
)

fig.show()
```

Being either at the beginning or at the end of the night seems to be better positions that being in the middle. Indeed, the positive answer rates are slightly lower in an intermediate position.
